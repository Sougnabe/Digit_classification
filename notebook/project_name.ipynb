# Notebook - Image Classification (Simple Version)This notebook shows the main steps: preprocessing, training, evaluation, and prediction.

```python
import sysfrom pathlib import Pathrepo_root = Path.cwd()if not (repo_root / 'src').exists():    repo_root = repo_root.parentif str(repo_root) not in sys.path:    sys.path.insert(0, str(repo_root))import osos.chdir(repo_root)import numpy as npimport pandas as pdimport matplotlib.pyplot as pltfrom sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, log_loss, classification_report, confusion_matrixfrom src.preprocessing import get_image_datasetsfrom src.model import build_model
```

```python
x_train, y_train, x_test, y_test, class_names = get_image_datasets(train_dir='data/train', test_dir='data/test', image_size=(64, 64))print('x_train:', x_train.shape)print('x_test :', x_test.shape)print('classes:', class_names)
```

## Preprocessing1. Read images from the class folders.2. Convert them to RGB.3. Resize them to 64x64.4. Normalize pixels between 0 and 1.5. Flatten the images for sklearn models.

```python
model = build_model(epochs=1)model.fit(x_train, y_train)y_pred = model.predict(x_test)y_proba = model.predict_proba(x_test)metrics = {    'accuracy': accuracy_score(y_test, y_pred),    'precision': precision_score(y_test, y_pred, average='weighted', zero_division=0),    'recall': recall_score(y_test, y_pred, average='weighted', zero_division=0),    'f1': f1_score(y_test, y_pred, average='weighted', zero_division=0),    'loss': log_loss(y_test, y_proba, labels=list(range(y_proba.shape[1]))),}pd.DataFrame([metrics])
```

```python
print(classification_report(y_test, y_pred, target_names=class_names, zero_division=0))cm = confusion_matrix(y_test, y_pred)plt.figure(figsize=(7, 5))plt.imshow(cm, cmap='Blues')plt.title('Confusion Matrix')plt.xlabel('Predicted')plt.ylabel('True')plt.colorbar()plt.show()
```

## Interpretation of 3 features1. Class distribution: an imbalance can bias the model.2. Image size: very different sizes can affect quality.3. Brightness: images that are too dark or too bright can change the prediction.Story: better balanced and more consistent data usually gives better performance.
